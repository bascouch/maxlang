## ORGINAL from dudas

```python
%matplotlib notebook
import numpy as np
import matplotlib.pyplot as plt 
import math as m

curve=1.

N = 1024
min_v = 0.
max_v = 1.

dy = max_v - min_v

input_array = np.linspace(0,1,num=N)

output_array = []

c1 = 1e-20
c2 = 1.2
c3 = .41
c4 = .91

hh = pow(((curve + c1) * c2), c3) * c4
if curve <0:
    hh = pow(((c1 - curve ) * c2), c3) * c4
    dy *= -1

ff = hh / (1 - hh)
eff = m.exp(ff) - 1
gh = (m.exp(ff * 0.5) - 1) / eff
bb = gh * (gh / (1 - (gh + gh)))
mm = 0.

if curve < 0:
    mm = 1 / (((m.exp(ff * (1/N)) - 1) / (eff * bb)) + 1)
    bb += 1
else:
    mm = ((m.exp(ff * (1/N)) - 1) / (eff * bb)) + 1

vv = bb
vv /= mm

for i in range(N):
    vv *= mm
    output_array.append((vv-bb)*dy +min_v)

plt.plot(output_array)
plt.show()
```

```python
%matplotlib notebook
import numpy as np
import matplotlib.pyplot as plt 
import math as m

curve=-.102

Nin=1024
min_v = 0.
max_v = 1.

dy = max_v - min_v

input_array = np.linspace(0,1,num=Nin)
output_array = []

c1 = 1e-20
c2 = 1.2
c3 = .41
c4 = .91

hh = pow(((curve + c1) * c2), c3) * c4
if curve <0:
    hh = pow(((c1 - curve ) * c2), c3) * c4
    dy *= -1

ff = hh / (1 - hh)
eff = m.exp(ff) - 1
gh = (m.exp(ff * 0.5) - 1) / eff
bb = gh * (gh / (1 - (gh + gh)))

if curve < 0:
    mm = 1 / (((m.exp(ff) - 1) / (eff * bb)) + 1)
    bb += 1
else:
    mm = ((m.exp(ff) - 1) / (eff * bb)) + 1
    

print(" ff "+str(ff)+"\n eff "+str(eff)+"\n gh "+str(gh)+"\n bb "+str(bb)+"\n mm "+str(mm))


## LOOP
output_array = ((bb * (np.power(mm,input_array)-1))*dy + min_v)

## PLOT
plt.plot(output_array)
plt.show()

print("OUT 0 : "+str(output_array[0])+" OUT 1 : "+str(output_array[Nin-1]))
```
